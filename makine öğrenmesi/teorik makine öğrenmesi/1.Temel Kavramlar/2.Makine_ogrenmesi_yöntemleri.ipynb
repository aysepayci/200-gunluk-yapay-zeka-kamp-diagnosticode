makine öğrenmesi terminolojisi

```python
#veri kümesi (Dataset) input-target(hedef)
#Öznitelik (Features) veri noktalarını tanımlar - modellenme ve tahmin için kullanılan değişkenler
#Etiketler (Labels) gözetimli öğrenmede hedef değişken- etiket özellikler karşı gelen gerçek çıktı..
```

```python
#model algoritmalar tarafından eğitilen veriler arasındaki örüntüyü tespit eden matamatiksel yapı- girdi alır çıktı verir -
#Eğitim (Training) maline öğrenimi modelinin dataset üzerindeki örnekler ile eğitilmesi süreci- genelleme yapılabilir hale gelir
#Doğrulama (Validation) eğitim sürecinde modelin performansını değerlendirir
#Tesy (Test) modelinin gerçek dünyada verikümesiyle test edilmesi
```

```python
#sınıflandırma (Classification) bir veri noktasını bir veya daha fazla sınıfa atayan tahmin problemidir. 
#Resgresyon (Regression) bir çıktı değişkenin sürekli bir değeri tahmin etme tahmin problemidir - regresyon çıktıları sayısal değerdir 
#Öğrenme Algoritmaları (Learning Algorithms) veri üzerinden örüntüleri keşfetmek ve modellemek için kullanılan matamatiksel ve istatiksel tekniklerdir
#Öznitelik mühendisliği (Feature Engineering) hücre Alan, hücre yarıçap, boyut - selection(fetures arasında seçme), extraction (yeni bir feature yapmak)
#Hiperparametreler (Hyperparameters) modelin yapılandırmasını ve eğitim sürecini kontrol eden parametrelerdir. 
```

```python
#overfitting ezberleme- modelin eğitim verilerine aşırı uyum sağlaması
#underfitting öğrenememe - eğitim verilerine uyum sağlayamama
#cross-Validation (çapraz doğrulama) modelin performansını değerlendirmek için teknik - veri kümesi birden fazla alt kümeye bölünür ve her bir alt küme sırasıyla eğitim ve doğrulama için kullanılır- bu sayede modelin genellleme yeteneği daha doğru şekilde değerlendirilmiş olur 
# Derin öğrenme (Deep Learing) yapay sinir ağları gibi karmaşık ve çok katmanlı modellerin kullanıldığı makine öğrenmesinin alt dalıdır - derin öğrenme büyük veri setleri üzerinde karmaşık örüntüleri öğrenmek için etkili yaklaşım
```
