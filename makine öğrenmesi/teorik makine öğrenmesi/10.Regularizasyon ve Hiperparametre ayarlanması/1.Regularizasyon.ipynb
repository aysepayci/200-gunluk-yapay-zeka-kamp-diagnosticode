L1 Regularizasyon (Lass Regularization)

```python
#L1 Regularizasyon, modelin katsayılarını (parametreleri) sıfıra yaklaştırmak için L1 normunu kullanır.
#Bu, bazı katsayıların tamamen sıfıra düşmesine neden olabilir, böylece model basitleşir ve daha iyi genelleme yapabilir.
```

L2 Regularizasyon (Ridye Regularizasyon)

```python
#L2 regularizasyonu,katsayıları sıfıra yaklaştırmak için karelerin toplamını (L2 normu) kullanır. Bu yöntem, büyük katsayıları azaltır ancak tamamen sıfıra düşmezler
#Model karmaşıklığını azaltır ve aşırı uyumu önler
```

L1 vs L2 arasındaki fark

```python
# öznitelik seçiminde L1 etkilidir
# Robustlık -dayanıklılık L1 etkilidirm
# Çözüm hızı L2 daha hızlı bazen L1 çözümünde türev tanımsız çıkabilir
```
