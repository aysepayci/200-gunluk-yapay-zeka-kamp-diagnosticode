öznitelik nedir?

```python
#öznitelik (feature), bir veri örneğini tanımlayan veya açıklayan herhangi bir özelliktir. 
#öznitelik, veri setindeki her bir gözlemin bir parçası oluşturan ölçülebilir veya gözlenebilir özelliklerdir
#öznitelikler, veri setinin her bir sütununu temsil eder. 
```

```python
#her bir sütun değerine öznitelik değeriz
```

Öznitelik mühendisliği

```python
#öznitelik mühendisliği (feature engineering), veri bilimi ve makine öğrenimi alanında oldukça kullanılan bir tekniktir 
# ve veri setindeki öznitelikleri (features) optimize etmek veya yeni öznitelik oluşturmak için yapılan işlemleri ifade eder
```

Öznitelik mühendisliğini önemi

```python
#model performansını artırma
#boyut azaltma - gereksiz veriyi çıkarma veya birleştirme korele=ilişkili modelin hızlı eğitilmesi 
#hesaplama maliyeylerini azaltma 
```

öznitelik seçimi

```python
#öznitelik seçimi (feature selection), veri setindeki bulunan öznitelik arasında en önemli veya en bilgilendirici olanları seçme veya belirleme sürecidir.
#Bu süreçte, veri setindeki gereksiz, terarlayan veya düşük etkili öznitelik tanımlanması ve çıkarılması hedeflenir
#Amacı, modelin performansını artırmak, karmaşıklığı azaltmak ve aşırı uydurma engellemektir
```

```python
#müşteri id çıkarıcam 
```

öznitelik seçimi yöntemleri

```python
#informasyon tabanlı yöntemler 
#wrapper yöntemleri - bir modelin performansına göre
#gömülü yöntemler - en önemli öznitelikleri belirlerler
#filter yöntemler - öznitelikler arasındaki ilişkiyi ona göre önemli olanı seçeriz
```

öznitelik çıkarma

```python
#öznitelik çıkarma (feature extraction), veri setindeki özniteliklerin temsilini değiştirme veya dönüştürme sürecidir. 
#bu süreçte, var olan öznitelikleri birleştirmesi, dönüştürülmesi veya yeni özniteliklerin oluşturulması yoluyla verinin temsilinin iyileştirilmesi amaçlanır. 
#öznitelik çıkarma, veri boyutunu azaltarak, gereksiz bilgiyi çıkararak ve verinin daha anlamlı bir şekilde temsil edilmesini sağlayarak model performansını artırabilir. 
```

öznitelik çıkarmanın önemi

```python
#model performansını arttırma
#boyut azaltma
#gereksiz bilgiyi çıkarma
#anlamlı temsilin sağlanması
```
