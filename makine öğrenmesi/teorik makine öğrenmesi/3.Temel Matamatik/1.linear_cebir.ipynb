```python
#Skaler (scaler)
#büyüklüğü (magnitude) belirten ve yalnızca bir sayıdan oluşan matamatiksel bir nesnedir. 
#yani, sadece büyüklüğü vardır ve bir yöne veya konuma sahip değildir
#bir otomobilin hızı, bir kişinin yaşını, veya bir oda sıcaklığı birer skaler değerdir.
#Vektör(vector)
#hem büyüklüğü (magnitude) hem de yönü (direction) belirten matematiksel bir nesnedir. bir başlangıç noktasından bir bitiş noktasına yönelen bir ok olarak düşünebilir. 
#bir otomobilin hızı ve yönü(örneğin, 50km/saat doğuya) birer vektördür. bir kuvvetin uygulandığı yön ve şiddeti de bir vektör olarak ifade edilebilir.  
```
