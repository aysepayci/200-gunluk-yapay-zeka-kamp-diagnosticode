olasılık dağılımları

```python
#normal dağılım (gauss dağılı)
#bernoulli dağılımı
#binom dağılımı
#poisson dağılımı
```

```python
# 1. Normal Dağılım (Gauss Dağılımı)
# Tanım: Normal dağılım, sürekli bir olasılık dağılımıdır ve birçok doğal olayın dağılımını tanımlar. Çan eğrisi şeklindedir.
# Özellikleri:
# Ortalamaya göre simetriktir.
# En yüksek olasılık ortalamada görülür.
# Parametreler: Ortalama (μ) ve standart sapma (σ).
# Örnek: Boy uzunlukları, sınav notları, kan basıncı ölçümleri.
```

```python
# 2. Bernoulli Dağılımı
# Tanım: Bernoulli dağılımı, yalnızca iki olası sonucu (Başarı - 1 veya Başarısızlık - 0) olan tek bir denemeyi temsil eder.
# Özellikleri:
# Deneme sayısı n = 1’dir.
# Parametre: Başarı olasılığı (p).
# Örnek: Yazı tura atma, bir hastalığın var olup olmaması.
```

```python
# 3. Binom Dağılımı
# Tanım: Binom dağılımı, n bağımsız Bernoulli denemesinin toplam başarılı sonuçlarını temsil eder.
# Özellikleri:
# Parametreler: Deneme sayısı (n) ve başarı olasılığı (p).
# Örnek: 10 kez yazı tura atıp kaç kez yazı geldiğini saymak.
```

```python
# 4. Poisson Dağılımı
# Tanım: Poisson dağılımı, belirli bir zaman veya alanda meydana gelen nadir olayların sayısını modellemek için kullanılır.
# Özellikleri:
# Parametre: Ortalama olay sayısı λ (lambda).
# Örnek: Bir hastaneye acil servise gelen hasta sayısı, bir web sitesine dakikada gelen ziyaretçi sayısı.
```
