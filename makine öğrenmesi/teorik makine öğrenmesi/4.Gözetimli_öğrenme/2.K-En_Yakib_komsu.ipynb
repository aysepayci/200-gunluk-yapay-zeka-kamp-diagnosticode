Durum çalışması

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Veriler
x1 = [1, 2, 1, 3, 4, 3]
x2 = [1, 1, 2, 3, 3, 4]
labels = ['A', 'A', 'A', 'B', 'B', 'B']

# Veri çerçevesi oluştur
import pandas as pd
df = pd.DataFrame({'x1': x1, 'x2': x2, 'label': labels})

# Grafik çizimi
plt.figure(figsize=(6, 4))
sns.scatterplot(data=df, x='x1', y='x2', hue='label', style='label', s=100, palette={'A': 'blue', 'B': 'red'})

# Noktalara etiket ekleme
for i in range(len(df)):
    plt.text(df.x1[i] + 0.1, df.x2[i] + 0.1, df.label[i], fontsize=12, color='black')

# Eksen ve başlık
plt.xlabel('X1 Değerleri')
plt.ylabel('X2 Değerleri')
plt.title('X1 ve X2 Arasındaki Dağılım')

# Grafiği göster
plt.grid(True)
plt.show()
```
