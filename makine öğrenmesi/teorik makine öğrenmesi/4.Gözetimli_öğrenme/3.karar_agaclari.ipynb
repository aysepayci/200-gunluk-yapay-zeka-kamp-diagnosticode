Bilgi kazancı (Information Gain)

```python
# bir özelliğin bölünme noktasının ne kadar iyi olduğunu belirlemek için kullanlılır
# özü, bir özelliğin seçilmesinin, veri kümesinin ne kadar iyi sınıflandırılcağına dair belirsizliği azaaltıp azaltmadığını ölçmektir.
# yani, bir özellik seçildiğindei yeni oluşan alt kümelerin daha homejen olup olmadığını değerlendirir
# bilgi kazancı ne kadar yüksekse, bir özellik o kadar iyi bir bölünme kriteri olarak kabul edilir
```

Entropi

```python
# #karar ağaçları (decision trees) içinde etropi, belirli bir düğümdeki veri noktalarının sınıflarının ne kadar karışık olduğunu ölçen bir metriktir.
# daha düşük bir entropi değeri, bir düğümün daha homojen sınıflara sahip olduğunu, yani veri noktalarının çoğunluğunun aynı sınıfa ait olduğunu gösterir.
# daha yüksek bir entropi değeri ise, bir düğümün daha karışık sınıflara sahip olduğunu, yani veri noktalarının farklı sınıflara daha eşit olarak dağıldığını gösterir
```
