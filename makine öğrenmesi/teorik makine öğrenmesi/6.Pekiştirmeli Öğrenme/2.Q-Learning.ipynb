```python
"""
1. Monte Carlo Methods (Monte Carlo Yöntemleri)
Tanım:
Monte Carlo yöntemleri, bir ortamda deneyimlerden örnekler toplayarak ve bunları istatistiksel olarak analiz ederek eylem-değer fonksiyonlarını (Q-fonksiyonu) tahmin etmeye çalışan yaklaşımlardır. Model tabanlı değil, örnekleme tabanlıdır ve bir duruma (state) ait değeri bulmak için birçok simülasyon çalıştırılır.

Örnek:
Bir satranç oyuncusu, belirli bir hamleyi yaptıktan sonra oyunun sonucunu birçok farklı simülasyonla test ederse ve hamlesinin uzun vadeli getirisini (kazanma olasılığını) hesaplamaya çalışırsa, Monte Carlo yöntemini kullanmış olur."""
"""
```

```python
"""
2. SARSA (State-Action-Reward-Action)
Tanım:
SARSA, önceden belirlenmiş bir politika (policy) kullanarak öğrenen ve güncellenmiş eylem-değer fonksiyonlarını hesaplayan bir yöntemdir. Gelecekteki eylemi de hesaba kattığı için adı SARSA (State-Action-Reward-State-Action) olarak anılır.
Öğrenme kuralı:
"""
#𝑄(𝑠,𝑎)←𝑄(𝑠,𝑎)+𝛼[𝑅+𝛾𝑄(𝑠′,𝑎′)−𝑄(𝑠,𝑎)]
"""
Burada:
s = Mevcut durum (state)
a = Seçilen eylem (action)
R = Alınan ödül (reward)
s′= Sonraki durum (next state)
a′= Sonraki eylem (next action)

Örnek:
Bir robotun oda içinde hareket ettiğini düşünelim. SARSA yöntemi, robotun mevcut hareketlerini ve ödüllerini değerlendirirken, gelecekte hangi yöne gitmesi gerektiğini de dikkate alarak öğrenmesini sağlar."""
```

```python
"""
3. Policy Gradient Methods (Politika Gradyan Yöntemleri)
Tanım:
Politika gradyan yöntemleri, doğrudan bir politika fonksiyonunu öğrenerek ve gradyan inişi kullanarak en iyi politikayı optimize etmeye çalışır. Klasik Q-learning gibi yöntemlerden farklı olarak, politika gradyan yöntemleri sürekli aksiyon alanlarında daha iyi çalışabilir.

Örnek:
Bir robotun yürüme şeklini öğrenmesini düşünelim. Politika gradyan yöntemleri, robota adım uzunluğu ve hızını sürekli olarak değiştirme esnekliği sağlar. Diskret (ayrık) aksiyon alanları yerine sürekli aksiyonlar kullanıldığı için, daha pürüzsüz bir yürüyüş elde edilir.

"""
```

```python
"""
4. Actor-Critic Methods (Aktör-Kritik Yöntemleri)
Tanım:
Bu yöntem Politika Gradyanları (Policy Gradient) ve Q-learning’i birleştirir.

Actor (Aktör): Politika fonksiyonunu günceller ve aksiyon seçer.
Critic (Kritik): Aktörün seçtiği aksiyonları değerlendirerek geri bildirim verir.
Örnek:
Bir oyun karakterinin düşmanlara saldırma stratejisini öğrendiğini düşünelim.

Actor, hangi saldırıyı yapması gerektiğine karar verir.
Critic, bu saldırının iyi mi kötü mü olduğunu değerlendirerek Actor'a geri bildirim verir.
"""
```

```python
"""
5. Temporal Difference Learning (TD Learning)
Tanım:
TD öğrenme, Monte Carlo ile Dinamik Programlamayı birleştirerek gelecekteki ödülleri tahmin etmeye çalışır. Q-learning ve SARSA gibi algoritmalar TD öğrenme ailesine aittir.

Örnek:
Bir hisse senedi yatırımcısı, hisselerinin değerini tahmin etmek istiyor.

Monte Carlo yaklaşımı: Gün sonunda hisselerin değerini hesaplar.
TD Learning yaklaşımı: Gün ortasında bir tahminde bulunur ve gün sonunda tahminini günceller."""
```

```python
"""
6. Thompson Sampling
Tanım:
Thompson Sampling, çok kollu bandit (multi-armed bandit) probleminde kullanılan bir yöntemdir. Belirsizliğe dayalı keşif (exploration) ve kazanç sağlamaya yönelik sömürü (exploitation) arasında denge kurar.
Her eylem için bir olasılık dağılımı oluşturur ve bu dağılımdan rastgele bir örnek çekerek aksiyon seçer.

Örnek:
Bir e-ticaret sitesi, farklı reklamları denemek istiyor. Thompson Sampling, hangi reklamın en fazla tıklanma oranına sahip olduğunu keşfetmek için farklı reklamları olasılıklara dayalı olarak gösterir ve en iyi seçeneği bulmaya çalışır.

"""
```

```python
"""
7. Upper Confidence Bound (UCB)
Tanım:
UCB, çok kollu bandit problemlerinde kullanılan başka bir keşif-sömürü yöntemidir.
Her eylemin güven aralığını hesaplar ve daha az keşfedilen eylemleri önceliklendirir.

Burada:
Q(a) → Seçilen aksiyonun beklenen ödülü
t → Toplam deneme sayısı
N(a) → Seçilen aksiyonun denenme sayısı
c → Keşif katsayısı

Örnek:
Bir restoran, farklı menü öğelerini tanıtmak istiyor.
UCB yöntemi, başlangıçta her yemeği bir kez deneyecek.
Daha sonra en başarılı olanları daha sık önererek müşteri memnuniyetini maksimize edecek."""
```
