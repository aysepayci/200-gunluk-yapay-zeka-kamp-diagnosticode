Olasılıksal

```python
# Bilgi kriteri, istatstiksel prosedürlerin etkinliği değerlendirmek için kullanılan bir tür olasılıksal ölçümdür.
# Yöntemleri, En Büyük Olabilirlik Tahmini'nin (MLE) log-olabilirlik çerçevesini kullanarak en etkili aday modelleri seçen bir puanlama sistemini içerir
#Örnekleme sadece model performansına odaklanırken, olasılıksal modelleme hem model performansına hem de karmaşıklığa odaklanır
#Karmaşıklık derecesini ve belirli bir modelin veri kümesine ne jkadar iyi uyduğu hesaplamak için üç istatistiksel yntem vardır:
#    Akaike Bilgi Kriteri(Akaike information Criteron-AIC)
#    Minimum Açıklama uzunluğu (Minimum Description Lenght-MDL)
#    Bayesian Bilgi Kriteri(Bayesian information Criterion -BIC)
```

```python
#Bölme işlemi rasgele veya zamana dayalı olabilir
#Rasgele bölme yöntemi, eğitim verisini eğitim, test ve doğrulama kümelerine rasgele böler.
#Bu işlem, modelin performansını çesşitli test kümelerinde kontrol etmek ve güvenirliğini gözlemlemek için tekrarlanır.
```

Örnekleme yöntemleri

```python
#Örnekleme yöntemleri, modelin eğitim almadığı veri örnekleri üzerinde performansını görmek için veri örnekleri yeniden düzenlemenin basit yöntemleridir.
#Başka bir deyişle, örnekleme, modelin genelleme yeteneğini belirlememizi sağlar.
#Örnekleme yöntemleri:
#   Bootstrap
#   Çarpraz doğrulama
#       K-Ford Çapraz Doğrulama
#       Leave-One-Out (LOO) Çapraz Doğrulama
```

Bootstrap

```python
#Bu teknik, bir veri setinden rasgele örneklemler çekmek suretiyle tekrarlanan örnekleme yaparak, popülasyon üzerinde istatistiksel sonuçların tahmin edilmesini sağlar.
#Bootstrap yöntemi, genelikle küçük veri ssetleri veya sınırlı veriye sahip durumlarda parametrik olmayan istatistiklerin güven aralıklarını ve standart hatalatını hesaplamak için kullanılır.
#   Veri setinden rastgele örneklemler çekilir
#   örneklemler üzerinde istatistikler hesaplanır 
#   Bu adımlar birçok kez tekrarlanır
```

K-Ford Çapraz Doğrulama

```python
# veri seti k eşit parçaya bölünür(katman)
# Ardından, her bir katman sırasıyla test seti olarak kullanılırken, diğer katmanlar birliştirilerek eğitim seti oluşturulur
# Bu işlem kere takrarlanır ve her bir katmanın test seti üzerindeki performansı değerlendirilir
# Sonuçlar genelikle ortalamalar alınarak rapor edilir
```

Leave-One-Out (LOO) Çapraz Doğrulama

```python
#Her bir veri noktası sırayla test seti olarak ayrılırken, geri kalan veri noktaları eğitim seti olarak kullanılır.
#Bu işlem veri noktalarının tamamı için tekrarlanır ve her bir veri noktasının test seti üzerindeki performansı değerlendirilir
```
